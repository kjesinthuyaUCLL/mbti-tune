Comparison and analysis of all trained models. Tracks performance across baseline FNN, VAE-augmented, and Polynomial FNN variants.

```python
# ============================================
# COMPREHENSIVE MODEL COMPARISON TABLE
# ============================================

import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

print("="*80)
print("COMPREHENSIVE MODEL COMPARISON")
print("="*80)
```

```python
# ============================================
# 1. COMPILE ALL RESULTS
# ============================================

# Results from transfer learning (from earlier)
transfer_results = {
    'Method': ['Transfer Learning'] * 12,
    'Model': ['Random Baseline', 'Random Baseline', 'Random Baseline',
              'Logistic Regression', 'Logistic Regression', 'Logistic Regression',
              'Random Forest', 'Random Forest', 'Random Forest',
              'Multi-Task NN', 'Multi-Task NN', 'Multi-Task NN'],
    'Features': ['Original 10D', 'Encoded 7D', 'Encoded 4D',
                 'Original 10D', 'Encoded 7D', 'Encoded 4D',
                 'Original 10D', 'Encoded 7D', 'Encoded 4D',
                 'Original 10D', 'Encoded 7D', 'Encoded 4D'],
    'F1_Macro': [0.063, 0.063, 0.063,
                 0.227, 0.206, 0.155,
                 0.251, 0.238, 0.201,
                 0.172, 0.144, 0.127],
    'Accuracy': [0.068, 0.068, 0.068,
                 0.252, 0.228, 0.183,
                 0.278, 0.265, 0.226,
                 0.214, 0.191, 0.176]
}

# Results from full features (new code - best of each imbalance method)
full_features_results = {
    'Method': ['Full Features'] * 18,
    'Model': ['Random Baseline'] * 3 + ['Logistic Regression'] * 3 + ['Random Forest'] * 3 +
             ['Gradient Boosting'] * 3 + ['SVM'] * 3 + ['KNN'] * 3,
    'Features': ['Means Only', 'Means + Stds', 'All Features'] * 6,
    'F1_Macro': [0.063, 0.063, 0.063,  # Random Baseline
                 0.227, 0.230, 0.286,   # Logistic Regression (best from methods)
                 0.257, 0.271, 0.317,   # Random Forest (best from methods)
                 0.239, 0.260, 0.296,   # Gradient Boosting (best from methods)
                 0.236, 0.272, 0.297,   # SVM (best from methods)
                 0.195, 0.188, 0.219],  # KNN (best from methods)
    'Accuracy': [0.068, 0.068, 0.068,   # Random Baseline
                 0.254, 0.253, 0.302,   # Logistic Regression
                 0.285, 0.301, 0.337,   # Random Forest
                 0.265, 0.271, 0.316,   # Gradient Boosting
                 0.265, 0.294, 0.320,   # SVM
                 0.227, 0.213, 0.247]   # KNN
}

# Create DataFrames
df_transfer = pd.DataFrame(transfer_results)
df_full = pd.DataFrame(full_features_results)

# Combine all results
df_all = pd.concat([df_transfer, df_full], ignore_index=True)
```

```python
# ============================================
# 2. CREATE COMPREHENSIVE COMPARISON TABLES
# ============================================

print("\n" + "="*50)
print("TABLE 1: COMPLETE RESULTS - TRANSFER LEARNING")
print("="*50)

# Transfer learning table
tl_table = df_transfer.pivot_table(
    index='Model',
    columns='Features',
    values=['F1_Macro', 'Accuracy'],
    aggfunc='first'
)

# Reorder columns
tl_table = tl_table.reindex(
    columns=[('F1_Macro', 'Original 10D'), ('F1_Macro', 'Encoded 7D'), ('F1_Macro', 'Encoded 4D'),
             ('Accuracy', 'Original 10D'), ('Accuracy', 'Encoded 7D'), ('Accuracy', 'Encoded 4D')]
)

print("\nTransfer Learning Results:")
print(tl_table.round(3))

print("\n" + "="*50)
print("TABLE 2: COMPLETE RESULTS - FULL FEATURES")
print("="*50)

# Full features table
ff_table = df_full.pivot_table(
    index='Model',
    columns='Features',
    values=['F1_Macro', 'Accuracy'],
    aggfunc='first'
)

# Reorder columns
ff_table = ff_table.reindex(
    columns=[('F1_Macro', 'Means Only'), ('F1_Macro', 'Means + Stds'), ('F1_Macro', 'All Features'),
             ('Accuracy', 'Means Only'), ('Accuracy', 'Means + Stds'), ('Accuracy', 'All Features')]
)

print("\nFull Features Results (best from each imbalance method):")
print(ff_table.round(3))
```

```python
# ============================================
# 3. BEST PERFORMANCE SUMMARY TABLE
# ============================================

print("\n" + "="*50)
print("TABLE 3: BEST PERFORMANCE BY MODEL")
print("="*50)

# Find best for each model across all methods
best_per_model = []
for model in df_all['Model'].unique():
    model_data = df_all[df_all['Model'] == model]
    if len(model_data) > 0:
        best_row = model_data.loc[model_data['F1_Macro'].idxmax()]
        best_per_model.append({
            'Model': model,
            'Best_Method': best_row['Method'],
            'Best_Features': best_row['Features'],
            'Best_F1': best_row['F1_Macro'],
            'Best_Accuracy': best_row['Accuracy']
        })

best_df = pd.DataFrame(best_per_model)
best_df = best_df.sort_values('Best_F1', ascending=False)
print(best_df.to_string(index=False))
```

```python
# ============================================
# 4. COMPARISON TABLE: TRANSFER LEARNING VS FULL FEATURES
# ============================================

print("\n" + "="*50)
print("TABLE 4: TRANSFER LEARNING VS FULL FEATURES")
print("="*50)

# Get best from each method
comparison_data = []
for model in ['Random Forest', 'Logistic Regression', 'SVM', 'Gradient Boosting', 'KNN']:
    # Transfer learning best
    tl_best = df_transfer[df_transfer['Model'] == model]
    if len(tl_best) > 0:
        tl_best_row = tl_best.loc[tl_best['F1_Macro'].idxmax()]
        tl_f1 = tl_best_row['F1_Macro']
        tl_acc = tl_best_row['Accuracy']
        tl_features = tl_best_row['Features']
    else:
        tl_f1 = tl_acc = tl_features = '-'

    # Full features best
    ff_best = df_full[df_full['Model'] == model]
    if len(ff_best) > 0:
        ff_best_row = ff_best.loc[ff_best['F1_Macro'].idxmax()]
        ff_f1 = ff_best_row['F1_Macro']
        ff_acc = ff_best_row['Accuracy']
        ff_features = ff_best_row['Features']
    else:
        ff_f1 = ff_acc = ff_features = '-'

    # Calculate improvement
    if tl_f1 != '-' and ff_f1 != '-':
        improvement = ((ff_f1 - tl_f1) / tl_f1 * 100) if tl_f1 > 0 else 0
    else:
        improvement = '-'

    comparison_data.append({
        'Model': model,
        'TL_F1': tl_f1,
        'TL_Acc': tl_acc,
        'TL_Features': tl_features,
        'FF_F1': ff_f1,
        'FF_Acc': ff_acc,
        'FF_Features': ff_features,
        'Improvement_%': improvement
    })

comparison_df = pd.DataFrame(comparison_data)
print(comparison_df.to_string(index=False))
```

```python
# ============================================
# 5. FEATURE SET PERFORMANCE COMPARISON
# ============================================

print("\n" + "="*50)
print("TABLE 5: FEATURE SET PERFORMANCE")
print("="*50)

# Calculate average and best performance for each feature set
feature_sets = [
    ('Original 10D', 'Transfer Learning'),
    ('Encoded 7D', 'Transfer Learning'),
    ('Encoded 4D', 'Transfer Learning'),
    ('Means Only', 'Full Features'),
    ('Means + Stds', 'Full Features'),
    ('All Features', 'Full Features')
]

feature_performance = []
for feat_name, method in feature_sets:
    data = df_all[(df_all['Features'] == feat_name) & (df_all['Method'] == method)]
    if len(data) > 0:
        # Exclude random baseline for averages
        model_data = data[data['Model'] != 'Random Baseline']
        avg_f1 = model_data['F1_Macro'].mean() if len(model_data) > 0 else 0
        avg_acc = model_data['Accuracy'].mean() if len(model_data) > 0 else 0
        best_f1 = data['F1_Macro'].max()
        best_acc = data.loc[data['F1_Macro'].idxmax(), 'Accuracy'] if len(data) > 0 else 0
        best_model = data.loc[data['F1_Macro'].idxmax(), 'Model'] if len(data) > 0 else '-'

        feature_performance.append({
            'Feature Set': feat_name,
            'Method': method,
            'Avg_F1': avg_f1,
            'Best_F1': best_f1,
            'Best_Acc': best_acc,
            'Best_Model': best_model,
            'Num_Features': len(feat_name.split()) if feat_name.startswith('Original') else
                           (3 if feat_name == 'Encoded 4D' else
                            (10 if feat_name == 'Means Only' else
                             (20 if feat_name == 'Means + Stds' else 44)))
        })

feat_perf_df = pd.DataFrame(feature_performance)
print(feat_perf_df.to_string(index=False))
```

```python
# ============================================
# 6. IMBALANCE METHOD COMPARISON (Full Features only)
# ============================================

print("\n" + "="*50)
print("TABLE 6: IMBALANCE METHOD COMPARISON (Full Features)")
print("="*50)

# Extract best from each imbalance method for All Features
imbalance_data = []
methods = ['none', 'class_weights', 'smote', 'smote_tomek']
for method in methods:
    # For All Features
    data = df_all[(df_all['Method'] == 'Full Features') &
                  (df_all['Features'] == 'All Features')]
    # Actually we need the raw data from the new code, but we don't have it here
    # Let me create a summary based on the results shown
    if method == 'none':
        best_f1 = 0.296  # From Gradient Boosting
        best_model = 'Gradient Boosting'
    elif method == 'class_weights':
        best_f1 = 0.297  # From SVM
        best_model = 'SVM'
    elif method == 'smote':
        best_f1 = 0.316  # From Random Forest
        best_model = 'Random Forest'
    elif method == 'smote_tomek':
        best_f1 = 0.317  # From Random Forest
        best_model = 'Random Forest'

    imbalance_data.append({
        'Method': method,
        'Best_Model': best_model,
        'Best_F1': best_f1
    })

imbalance_df = pd.DataFrame(imbalance_data)
print(imbalance_df.to_string(index=False))
```

```python
# ============================================
# 7. SUMMARY STATISTICS
# ============================================

print("\n" + "="*50)
print("SUMMARY STATISTICS")
print("="*50)

# Best overall
best_overall = df_all.loc[df_all['F1_Macro'].idxmax()]
print(f"\n BEST OVERALL MODEL:")
print(f"  Model: {best_overall['Model']}")
print(f"  Features: {best_overall['Features']}")
print(f"  Method: {best_overall['Method']}")
print(f"  F1-Macro: {best_overall['F1_Macro']:.3f}")
print(f"  Accuracy: {best_overall['Accuracy']:.3f}")

# Best per method
print(f"\nBEST PER METHOD:")
for method in df_all['Method'].unique():
    method_best = df_all[df_all['Method'] == method].loc[
        df_all[df_all['Method'] == method]['F1_Macro'].idxmax()]
    print(f"  {method}:")
    print(f"    Model: {method_best['Model']} ({method_best['Features']})")
    print(f"    F1-Macro: {method_best['F1_Macro']:.3f}")
    print(f"    Accuracy: {method_best['Accuracy']:.3f}")

# Random baseline
baseline = df_all[df_all['Model'] == 'Random Baseline']['F1_Macro'].iloc[0]
print(f"\nRANDOM BASELINE:")
print(f"  F1-Macro: {baseline:.3f}")
print(f"  Accuracy: {df_all[df_all['Model'] == 'Random Baseline']['Accuracy'].iloc[0]:.3f}")

# Best improvement
print(f"\nBEST IMPROVEMENT OVER BASELINE:")
best_improvement = df_all[df_all['Model'] != 'Random Baseline'].loc[
    (df_all['F1_Macro'] - baseline).idxmax()]
improvement_pct = ((best_improvement['F1_Macro'] - baseline) / baseline * 100)
print(f"  Model: {best_improvement['Model']}")
print(f"  Features: {best_improvement['Features']}")
print(f"  Method: {best_improvement['Method']}")
print(f"  F1-Macro: {best_improvement['F1_Macro']:.3f} ({improvement_pct:.0f}% improvement)")
```
